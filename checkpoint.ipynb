```python
def calculatrice(num1, num2, operation) :
    if operation == 'add':
        resultat = num1 + num2
    elif operation == 'sub':
        resultat = num1 - num2
    elif operation == 'mul':
        resultat = num1 * num2
    elif operation == 'div':
        if num2 != 0:
            resultat = num1 / num2
        else:
            resultat = print("Erreur Division par zéro")
    else:
        resultat = "Erreur operation non valide"
    return resultat
calculatrice(25 , 25 , 'add')
```
